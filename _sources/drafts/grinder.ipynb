# Rensning af grinder

---

## 🧽 Sådan renser du en aluminium-grinder

---

### 1. Skil alt ad

- Skru alle dele fra hinanden (låg, midterdel, bund, evt. si).  
- Fjern eventuelle gummipakninger eller magneter, hvis de kan tages ud.


---

### 2. Fjern løst støv og skidt

- Brug en tør pensel, tandbørste eller vatpind til at fjerne så meget løst støv og harpiks som muligt.  
- Du kan evt. blæse med trykluft eller bare puste gennem delene.


---

### 3. Rens med alkohol

Den bedste måde at rense en gammel grinder på (især aluminium) er **isopropylalkohol (IPA) 90–99 %**.  
Sådan gør du:

1. Læg delene i en skål eller et glas med isopropylalkohol.  
2. Lad dem ligge i **10–15 minutter** (ikke for længe – så undgår du at angribe farve eller belægning).  
3. Brug en tandbørste eller vatpind til at skrubbe alle hjørner og tænder.  
4. Skyl grundigt i **varmt vand** bagefter.  
5. Tør helt med papir eller lad dem **lufttørre 100 %**.

💡 Hvis du ikke har isopropylalkohol, kan du bruge **klar eddike** eller **sæbevand** — men alkohol er klart bedst til harpiks og gammelt snavs.


---

### 4. (Valgfrit) Polér lidt

Hvis grinderen er blevet **mat** eller “**ru**” efter mange års brug, kan du gøre overfladen glat igen:

- Dyp en klud i **en smule madolie** (fx olivenolie).
- Gnid let på overfladen.
- Tør grundigt alt overskydende olie af bagefter.

➡️ Det fjerner **oxidation** og gør metallet **glat og blank** igen.  
⚠️ Brug kun **meget lidt olie**, og sørg for at **tørre alt overskud af!**


---

### 5. Sørg for, at alt er helt tørt

Det er vigtigt, at **alle dele er fuldstændig tørre**, især ved gevind og magnet, da **fugt kan skabe ny oxidation**.

- Tør med **rent papir** eller klud.
- Lad gerne delene stå **åbent i 24 timer** et tørt sted, inden du samler og bruger dem igen.


---

## ⚠️ Undgå

Visse metoder og materialer kan skade din grinder – især hvis den er lavet af aluminium.

- **Stålbørster eller groft sandpapir**  
  → Kan ridse overfladen og føre til hurtigere oxidering.

- **Opvaskemaskine**  
  → Varmen og sæben kan beskadige farven og opløse lim i magneterne.

- **Blød plast- eller trægrinder + alkohol**  
  → Alkohol kan opløse materialet – brug kun alkohol til metalgrindere.


---
