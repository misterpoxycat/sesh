# Rygning vs. vaping

Når man vil forstå forskellen mellem rygning og vaping af cannabis, handler det ikke kun om smag og oplevelse. **Temperatur, forbrænding, THC- og terpenoptagelse samt sundhedsrisici** spiller også en vigtig rolle.

Tabellen nedenfor giver et hurtigt overblik over, hvad du får med de to metoder, sat i relative tal, så det er nemt at sammenligne.

## 🌿💨 Rygning vs. vaping af cannabis

| Faktor | **Rygning (joint, pibe, bong)** | **Vaping (fx XMax V3 Pro, Mighty, Arizer)** |
|--------|----------------------------------|---------------------------------------------|
| 🔥 **Opvarmningsmetode** | Forbrænding med glød på **600–900 °C**. Foran gløden findes en **fordamperzone på 150–250 °C**, hvor lidt THC fordamper, før materialet brænder. | Kontrolleret fordampning ved **160–220 °C** – ingen forbrænding, kun damp. |
| 💨 **Indhold i damp/røg** | Indeholder **kulilte, tjære, PAH’er og ammoniak** fra forbrænding. | Indeholder kun **cannabinoider og terpener**, ingen forbrændingsbiprodukter. |
| 🌡️ **Temperaturkontrol** | Ingen – varmen svinger voldsomt afhængig af sug og glød. | Præcis styring i grader – stabil opvarmning. |
| 🌿 **THC-udnyttelse** | **40–70 %** – noget fordamper før gløden, men resten ødelægges i ilden. | **85–95 %** – næsten alt THC og terpener bevares. |
| 🧠 **High / effekt** | Hurtig, hård og “tung” rus, påvirket af CO og forbrændingsgasser. | Klarere, mere fokuseret og justerbar oplevelse. |
| 👃 **Smag & aroma** | Brændt og skarp – de fleste terpener destrueres over 200 °C. | Frisk, aromatisk og terpenernes fulde profil bevares. |
| 😶‍🌫️ **Lungepåvirkning** | Høj irritation, hoste, slim og nedsat lungefunktion. | Meget lav – typisk **3–5 % af belastningen** ved rygning. |
| 💉 **Eksponering for toksiner** | Sæt som reference: **100 %** | **3–5 %** (90–97 % reduktion i tjære og CO). |
| ❤️ **Relativ sundhedsrisiko (samlet)** | **100 %** | **≈ 3–5 %** ved korrekt brug af kvalitetsvape. |
| 💧 **Lugt** | Kraftig, hænger i tøj og rum. | Mild og hurtigt forsvindende. |
| ⚙️ **Effektivitet / forbrug** | Meget spild – THC brænder væk mellem sugene. | 30–40 % bedre udnyttelse af materiale. |
| ♻️ **Efterladenskaber** | Aske, filter, papir. | Kun brun “vaped” urt (kan bruges i edibles). |
| 💨 **Fysiologisk varme** | Indåndet røg omkring **100–150 °C**. | Damp **50–90 °C** – langt mildere. |

---

### 🔬 Dokumentation

- *Gieringer et al., 2004 (MAPS)* – 88–99 % reduktion i toksiske forbindelser ved vaporisering.  
- *Van Dam & Earleywine, 2010 (Harm Reduction Journal)* – Vapere har 95 % færre luftvejssymptomer.  
- *Health Canada, 2022* – Tjæredannelse reduceret med 95 % ved 200 °C i laboratorietests.

---

### 🧠 Kort opsummeret

> Rygning = 100 % eksponering, halvdelen af THC’en går tabt.  
>
> Vaping = **3–5 % af sundhedsrisikoen** og **næsten dobbelt udnyttelse** af indholdet.  
>
> Renere, sundere og lettere at styre — *mere rus, mindre røg.* 💨
