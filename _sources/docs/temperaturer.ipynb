# Guide til temperaturer og effekter ved vaping af cannabis

---

## Setup (XMax V3 Pro + kapsel)
- XMax V3 Pro markedsføres som en konvektions-vaporizer. I praksis bidrager den opvarmede skål også med en vis konduktiv opvarmning, især under længere sessioner. Ved brug af en **doseringskapsel** ændres varmeoverførslen, fordi kapslen tilfører ekstra termisk masse og udgør et ekstra led mellem varmeelementet og urten. Som følge heraf kan urtematerialets temperatur være lavere end den temperatur, der vises på enheden ved identiske brugsforhold.
- I tabellerne nedenfor angiver "Reel fordampning (uden kapsel)" omtrentlige temperaturer fra litteraturen, hvor de enkelte forbindelser begynder at frigives eller fordampe i målbare mængder. "Enhedsindstilling (med kapsel)" er et vejledende skøn over den enhedsindstilling, der typisk giver en tilsvarende effekt ved brug af doseringskapsel. Da kapslen ændrer varmeoverførslen, findes der ikke én fast temperaturkorrektion, som gælder under alle forhold.

## 🌿 Cannabinoider og temperaturer

|   | Stof | Reel fordampning (uden kapsel) | Enhedsindstilling (med kapsel) | Effekt / high |
|------|--------------------------------|------------------------------------|----------------|------|
| 🟩 | **CBG (cannabigerol)** | ~125 °C | 135–145 °C | Mild energi og fokus, synergistisk med THC. |
| 🟧 | **THC (Δ⁹-tetrahydrocannabinol)** | ~157 °C | 165–175 °C | Hovedansvarlig for eufori, kreativitet og “head high”. Lavere temp = klarere, lettere high. |
| 🟦 | **CBD (cannabidiol)** | ~160–180 °C | 170–190 °C | Ikke-psykoaktivt, dæmper angst/paranoia, giver ro og balance. |
| 🟪 | **CBN (cannabinol)** | ~185 °C | 195–205 °C | Nedbrydningsprodukt af THC → mere søvnig og sedativ virkning. |
| 🔴 | **THCV** | ~220 °C | 230–240 °C | Stimulerende, appetitdæmpende, mere sjælden cannabinoid. |

## 🌸 Terpener og temperaturer

|---------|--------------------------|------------------------|------------------|------|
| 🌲 | **Pinene** | ~155 °C | 165–170 °C | Fyrretræ – klarhed, fokus, modvirker hukommelsestab fra THC |
| 🌿 | **Myrcen** | ~166 °C | 175–180 °C | Jord, moskus – muskelafslappende, søvnig, “stoned” følelse |
| 🍋 | **Limonen** | ~176 °C | 185–190 °C | Citrus – løftende, glad, antidepressiv |
| 💜 | **Linalool** | ~198 °C | 205–215 °C | Lavendel – beroligende, angstdæmpende |
| 🌶️ | **Caryophyllen** | ~200 °C | 210–220 °C | Krydret/peber – anti-inflammatorisk, virker også på CB₂-receptorer |

## ⚡ Zoner og temperaturer

|   | Zone | Reel fordampning (uden kapsel) | Enhedsindstilling (med kapsel) | Karakteristik |
|---|------|-------------------------------|-------------------------|----------------|
| 🧠 | **Klar & kreativ** | ~160–170 | 170–180 | Let, cerebral, frisk aroma. Primært THC + terpener (pinene, myrcen). |
| ⚖️ | **Balanceret** | ~175–185 | 185–195 | THC + CBD + limonen → klassisk “all-round” high med stadig aroma. |
| 🌙 | **Afslappende** | ~190–200 | 200–210 | Tungere, mere fysisk og rolig virkning. CBN + linalool dominerer. |
| 💤 | **Søvndyssende** | ~205–215 | 215–230 | Meget tung, terpener brænder hurtigt af, maksimal ekstraktion. |
