# Guide til temperaturer og effekter ved vaping af cannabis

```{admonition} Om denne side
:class: note

- Denne side giver et samlet **overblik** over, hvordan temperaturen påvirker en vaping-session.
- Først introduceres de overordnede **temperaturzoner**, hvorefter tabellerne viser, hvilke **cannabinoider** (aktive stoffer) og **terpener** (aromatiske forbindelser) der typisk bidrager til de forskellige zoner.
- Næste kapitel går mere i dybden med terpenernes egenskaber og deres betydning for aroma og oplevelse.
```

---

## Setup (XMax V3 Pro + kapsel)

XMax V3 Pro markedsføres som en konvektionsbaseret vaporizer. I praksis bidrager den opvarmede skål også med en vis konduktiv opvarmning, især under længere sessioner. Ved brug af en **doseringskapsel** ændres varmeoverførslen, fordi kapslen tilfører ekstra termisk masse og samtidig fungerer som et ekstra lag mellem varmeelementet og urten. Som følge heraf kan urtematerialets temperatur være lavere end den temperatur, der vises på enheden ved identiske brugsforhold.

I tabellerne nedenfor angiver **"Reel fordampning (uden kapsel)"** de omtrentlige temperaturer, hvor de enkelte forbindelser ifølge litteraturen begynder at frigives eller fordampe i målbare mængder. **"Enhedsindstilling (med kapsel)"** er et vejledende skøn over den enhedsindstilling, der typisk giver en tilsvarende effekt ved brug af doseringskapsel. Da kapslen ændrer varmeoverførslen, findes der ikke én fast temperaturkorrektion, som gælder under alle forhold.

Temperaturen er derfor den vigtigste faktor, når du vaporiserer cannabis. Den afgør ikke blot, hvor meget damp der produceres, men også hvilke cannabinoider og terpener der bidrager mest til oplevelsen. Det betyder, at den samme cannabis kan ændre karakter, efterhånden som temperaturen stiger.

```{admonition} Om temperaturintervallerne
:class: hint
**Temperaturintervallerne er vejledende og overlapper.** De angiver de temperaturer, hvor de enkelte cannabinoider og terpener typisk bidrager mest til oplevelsen. Ved højere temperaturer kan de fortsat bidrage til oplevelsen, men deres relative betydning ændrer sig, efterhånden som de frigives og gradvist udtømmes.
```

## ⚡ Zoner og temperaturer

|   | Zone | Reel fordampning (uden kapsel) (°C) | Enhedsindstilling (med kapsel) (°C) | Karakteristik |
|---|------|-------------------------------|-------------------------|----------------|
| 🧠 | **Klar & kreativ** | ~160–170 | 170–180 | Let, cerebral, frisk aroma. Primært THC + terpener (pinene, myrcen). |
| ⚖️ | **Balanceret** | ~175–185 | 185–195 | THC + CBD + limonen → klassisk “all-round” high med stadig aroma. |
| 🌙 | **Afslappende** | ~190–200 | 200–210 | Tungere, mere fysisk og rolig virkning. CBN + linalool dominerer. |
| 💤 | **Tung & fysisk** | ~205–215 | 215–230 | Meget kraftig og kropslig virkning med maksimal ekstraktion. De fleste terpener er nu fordampet eller nedbrudt. |

## 🌿 Cannabinoider og temperaturer

|   | Stof | Reel fordampning (uden kapsel) (°C) | Enhedsindstilling (med kapsel) (°C) | Effekt / oplevelse |
|------|--------------------------------|------------------------------------|----------------|------|
| 🟩 | **CBG (cannabigerol)** | ~125 | 135–145 | Forbindes med klarhed og fokus, synergistisk med THC. |
| 🟧 | **THC (Δ⁹-tetrahydrocannabinol)** | ~155 | 165–175 | **Primær psykoaktiv cannabinoid.** Hovedansvarlig for eufori, kreativitet og "head high". Lavere temperatur forbindes ofte med en klarere oplevelse. |
| 🟦 | **CBD (cannabidiol)** | ~160–180 | 170–190 | Forbindes med ro og balance og kan dæmpe nogle af THC's uønskede effekter. |
| 🟪 | **CBN (cannabinol)** | ~185 | 195–205 | Dannes ved nedbrydning af THC og forbindes ofte med en mere tung og afslappende oplevelse. |
| 🔴 | **THCV** | ~220 | 230–240 | Sjælden cannabinoid, der afhængigt af dosis kan opleves som mere stimulerende. |
