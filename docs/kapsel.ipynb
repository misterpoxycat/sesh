# Kapselguide – XMax V3 Pro

## 🪷 The Way of Master Sesh

Der findes mange måder at bruge temperatur i en vaporizer-session på. Nogle går direkte til et fast *sweet spot* for hurtig effekt, mens andre jagter maksimal udnyttelse ved høj temperatur fra start. Master Sesh vælger en tredje tilgang: en gradvis, kontrolleret opvarmning, hvor plantens ønskede nuancer får lov til at udfolde sig i deres eget tempo.

**Kernen i Master Sesh’ metode er ikke at ramme én rigtig temperatur, men at bevæge sig gennem ekstraktionskurven trin for trin.** Ved brug af en doseringskapsel og en langsom temperaturstigning fra 165 °C til 192 °C over 10 sug, opbygges sessionen som en kontrolleret progression fra lette, aromatiske topnoter til en målrettet ekstraktion af de aktive forbindelser.

De første sug ved lavere temperaturer handler primært om terpener og den tidlige *lift* i oplevelsen – den klare, lyse og aromatiske del af planten. Midt i forløbet, omkring 175–185 °C, træder THC-dominansen tydeligt frem og giver den ønskede mentale og energiske effekt. De sidste sug nær 185–192 °C fungerer som en afsluttende fase, hvor resten af materialet udnyttes uden at skubbe oplevelsen fuldt ind i tunghed eller *couch*-territorie.

Den oplevelse, Master Sesh søger, er hverken den letteste eller den tungeste. **Målet er harmoni.** En balanceret session, hvor aroma, mental klarhed og kropslig afslapning udvikler sig gradvist og understøtter hinanden. De tidlige terpener giver en frisk og levende begyndelse, THC bygger oplevelsen op gennem midten af sessionen, og de sidste sug tilfører en behagelig kropslig ro uden at bevæge sig over i den tunge, søvndyssende ende af spektret. Resultatet er en oplevelse, der bevarer både nærvær og funktionalitet, samtidig med at kroppen får lov til at slappe af. Effekten udvikler sig gradvist og opleves derfor ofte som klar og afbalanceret frem for eksplosiv eller tung.

**For Master Sesh er målet ikke maksimal intensitet, men maksimal sammenhæng i oplevelsen.** Fokus er på *entourage*-effekten – samspillet mellem terpener, THC og de øvrige cannabinoider – hvor de forbindelser, der bidrager til den ønskede oplevelse, får lov til at udfolde sig gradvist gennem sessionen.

Resultatet er en session, der ikke topper hurtigt og falder af, men i stedet bevæger sig i bølger: fra aromatisk klarhed, over funktionel energi, til en kontrolleret afslutning af ekstraktionsforløbet.

Master Sesh søger ikke maksimal kraft i hvert øjeblik. Han søger balance. En oplevelse, hvor også de mest flygtige nuancer får plads til at udvikle sig, uden at sessionen presses længere, end formålet kræver.

Som Master Sesh altid siger:

> *Det handler ikke om at ramme den rigtige temperatur.*  
> *Det handler om at følge den.*

## 📋 Kapselguide

### 1. Forbered urten

- Mal urten **mellemfin** (ikke støv, men heller ikke grove stykker).
- Den skal være **let sprød** – ikke fugtig, men heller ikke knastør.
  - For fugtig → mindre damp.
  - For tør → harsk smag og hurtigere opvarmning.

### 2. Fyld kapslen korrekt

- Fyld **50–70 %** – næsten fuld, men uden at presse.
- Jævn let overfladen med en fingerspids eller værktøj.
- Sørg for, at luften stadig kan passere gennem urten.

### 3. Tid

- Sæt session til **6 minutter**.
  - XMax V3 Pro har to sessions-længder: 4 og 6 minutter.

### 4. Forvarmning og temperatur

- Tryk start og indstil til **165 °C**.
- Når enheden vibrerer, er den nået den indstillede temperatur, og sessionen starter automatisk nedtælling.
- Vent **25 sekunder**.
  - Dette giver kapslen tid til at nå mere stabil driftstemperatur.

### 5. Sug og sessionsteknik

- Sug **langsomt og roligt i ca. 10 sekunder**.
  - Forestil dig, at du langsomt drikker varm te gennem sugerør.
  - For hurtige sug → kammeret køles ned, og dampen bliver tynd.

- Hæv temperaturen med **3 grader** og vent igen **ca. 25 sekunder**.

- Fortsæt cyklussen: sug **10 sekunder**, hæv **3 grader**, vent **25 sekunder**.

- Intensiteten udvikler sig i faser. I de højere 170’ere og ind i 180’erne falder smagen gradvist, mens THC-effekten fortsat bliver mere fremtrædende.

- Hvis du følger cyklussen, vil sidste sug typisk falde, når der er 20 sekunder tilbage på uret.

### 6. Afslutning og anbefaling

- Det anbefales at planlægge cyklussen, så du når at skrue temperaturen ned til **165 °C**, inden sessionen slutter.
- XMax V3 Pro starter ellers automatisk op på den temperatur, den stod på ved afslutning.
- Ved selv at skrue ned sikrer du en mere konsistent start på næste session og undgår, at enheden “husker” en høj sluttemperatur.

## ⚖️ Mere end én vej

Master Sesh' metode er ikke en facitliste, men et udgangspunkt. Den er udviklet til en balanceret session, hvor aroma, funktionel effekt og kontrolleret ekstraktion vægtes lige højt. Afhængigt af dit formål kan du justere metoden.

- **Mere smag og aroma:** Bliv længere ved 165–175 °C eller sænk starttemperaturen en smule.
- **Mere mental effekt:** Brug flere sug omkring 175–185 °C, hvor THC dominerer mest.
- **Mere kropslig afslapning:** Lad sessionen fortsætte lidt længere op mod 195–200 °C, hvor flere tungere forbindelser bidrager.
- **Maksimal ekstraktion:** Fortsæt helt op mod 210–220 °C. Det giver en mere komplet udnyttelse af materialet, men på bekostning af en del af de tidlige aromaer og den lette karakter, som Master Sesh-metoden søger at bevare.

### ⚡ Mere energi og mental klarhed

Master Sesh-metoden er udviklet til en afbalanceret oplevelse og afsluttes normalt ved **192 °C**. Hvis du ønsker at bevare en lettere og mere funktionel oplevelse – f.eks. før arbejde, studier eller andre opgaver, der kræver fokus – kan du vælge at stoppe temperaturstigningen lidt tidligere, f.eks. ved **186 °C**. Fortsæt herefter sessionen som normalt, men uden at hæve temperaturen yderligere.

På den måde tilbringes en større del af sessionen i temperaturintervallet, hvor den klare og funktionelle oplevelse typisk er mest fremtrædende, samtidig med at bidraget fra de forbindelser, der især frigives ved højere temperaturer, reduceres. Resultatet vil ofte være en oplevelse, der føles lettere, klarere og mindre kropslig end den fulde Master Sesh-session.

### 🌙 Mere kropslig afslapning

Master Sesh-metoden afsluttes som sagt normalt ved **192 °C**, hvor oplevelsen stadig bevarer en god balance mellem mental klarhed og kropslig ro. Hvis du ønsker en tungere og mere afslappende oplevelse, kan du vælge at starte sessionen ved en højere temperatur, f.eks. **175 °C** eller højere.

På den måde tilbringes en større del af sessionen i de temperaturintervaller, hvor de forbindelser, der typisk bidrager til en mere kropslig oplevelse, frigives. Resultatet vil ofte være en dybere afslapning og en oplevelse, der føles tungere og mindre energisk end den oprindelige Master Sesh-session.

Bemærk, at urten ekstraheres hurtigere ved højere temperaturer. Starter du f.eks. ved **190 °C**, vil hvert sug typisk frigive flere aktive forbindelser, hvilket typisk giver en hurtigere og mere intens opbygning af oplevelsen. Til gengæld vil sessionen ofte bestå af færre brugbare sug, før dampen begynder at smage brændt.
