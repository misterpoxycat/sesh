# Kapselguide – XMax V3 Pro

## 🪷 The Way of Master Sesh

Der findes mange måder at bruge temperatur i en vaporizer-session på. Nogle går direkte til et fast “sweet spot” for hurtig effekt, mens andre jagter maksimal udnyttelse ved høj temperatur fra start. **Master Sesh vælger en tredje tilgang: en gradvis, kontrolleret opvarmning, hvor hele spektret af plantens forbindelser får lov til at komme i spil over tid.**

Kernen i Master Sesh’ metode er ikke at ramme én perfekt temperatur, men at bevæge sig gennem ekstraktionskurven trin for trin. Ved brug af en doseringskapsel og en langsom temperaturstigning fra ca. 165 °C til 192 °C over 10 sug, opbygges sessionen som en kontrolleret progression fra lette, aromatiske topnoter til en målrettet ekstraktion af de aktive forbindelser.

De første sug ved lavere temperaturer handler primært om terpener og den tidlige “lift” i oplevelsen – den klare, lyse og aromatiske del af planten. Midt i forløbet, omkring 175–185 °C, træder THC-dominansen tydeligt frem og giver den ønskede mentale og energiske effekt. De sidste sug nær 185–192 °C fungerer som en afsluttende fase, hvor resten af materialet udnyttes uden at skubbe oplevelsen fuldt ind i tunghed eller “couch”-territorie.

For Master Sesh er målet ikke maksimal intensitet, men maksimal sammenhæng i oplevelsen. Fokus er på entourage-effekten – samspillet mellem terpener, THC og de øvrige cannabinoider – hvor de forbindelser, der bidrager til den ønskede oplevelse, får lov til at udfolde sig gradvist gennem sessionen.

Resultatet er en session, der ikke topper hurtigt og falder af, men i stedet bevæger sig i bølger: fra aromatisk klarhed, over funktionel energi, til en kontrolleret afslutning af ekstraktionsforløbet.

Master Sesh søger ikke maksimal kraft i hvert øjeblik. Han søger balance. En oplevelse, hvor også de mest flygtige nuancer får plads til at udvikle sig, uden at sessionen presses længere, end formålet kræver.

For Master Sesh handler det ikke om at ramme temperaturen.  
Det handler om at følge den.

## 📋 Kapselguide

### 1. Forbered urten

- Mal urten **mellemfin** (ikke støv, men heller ikke grove stykker).
- Den skal være **let sprød** – ikke fugtig, men heller ikke knastør.
  - For fugtig → mindre damp.
  - For tør → harsk smag og hurtigere opvarmning.

### 2. Fyld kapslen korrekt

- Fyld **50–70 %** – næsten fuld, men uden at presse.
- Jævn let overfladen med en fingerspids eller værktøj.
- Sørg for, at luften stadig kan passere gennem urten.

### 3. Tid

- Sæt session til **6 minutter**.
  - XMax V3 Pro har to sessions-længder: 4 og 6 minutter.

### 4. Forvarmning og temperatur

- Tryk start og indstil til **165 °C**.
- Når enheden vibrerer, er den nået den indstillede temperatur, og sessionen starter automatisk nedtælling.
- Vent **25 sekunder**.
  - Dette giver kapslen tid til at nå mere stabil driftstemperatur.

### 5. Sug og sessionsteknik

- Sug **langsomt og roligt i ca. 10 sekunder**.
  - Forestil dig, at du langsomt drikker varm te gennem sugerør.
  - For hurtige sug → kammeret køles ned, og dampen bliver tynd.

- Hæv temperaturen med **3 grader** og vent igen **ca. 25 sekunder**.

- Fortsæt cyklussen: sug **10 sekunder**, hæv **3 grader**, vent **25 sekunder**.

- Intensiteten udvikler sig i faser. I de højere 170’ere og ind i 180’erne falder smagen gradvist, mens THC-effekten fortsat bliver mere fremtrædende.

- Hvis du følger cyklussen, vil sidste sug typisk falde, når der er 20 sekunder tilbage på uret.

### 6. Afslutning og anbefaling

- Det anbefales at planlægge cyklussen, så du når at skrue temperaturen ned til **165 °C**, inden sessionen slutter.
- XMax V3 Pro starter ellers automatisk op på den temperatur, den stod på ved afslutning.
- Ved selv at skrue ned sikrer du en mere konsistent start på næste session og undgår, at enheden “husker” en høj sluttemperatur.
