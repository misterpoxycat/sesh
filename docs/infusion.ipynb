# 💡 Infusion af vapet urt i olie eller smør

### 1. Forbered urten
- Sørg for urten **ikke er for fugtig** – let tørret om nødvendigt.  
- Knus urten let for bedre kontakt med fedtet, men undgå pulverisering.  

### 2. Vælg fedtstof
- **Olie:** kokosolie eller olivenolie holder bedst på THC og THCV, tåler længere opbevaring.  
- **Smør:** fungerer også, men bør opbevares i køleskab og bruges hurtigere.  

### 3. Forhold urt til fedt
- Vapet urt er mindre potent → brug **4–5× mængde i forhold til frisk urt** for tilsvarende effekt.  
- Juster lidt afhængigt af ønsket rus (mere energi → bland med frisk urt).  

### 4. Infusionsmetode (lav varme)
- Kom fedtstof i gryde eller dobbeltkoger.  
- Tilsæt den vapede urt.  
- Hold varmen **lav (~80–100 °C)** i 1–2 timer.  
  - Rør lejlighedsvis, så urten fordeles jævnt.  
- Undgå kogning – THC nedbrydes ved >160 °C.  

### 5. Filtrering / presning
- Filtrer urten med **fin si eller ostelærred**.  
- **Pres urten** let med ske eller fingre for at få mest muligt olie ud.  
- Eventuelt filtrer igen for klar olie.  

### 6. Opbevaring
- **Glasbeholder med tætsluttende låg**, mørkt og køligt (køleskab anbefales).  
- Olie kan fryses i små portioner for længere holdbarhed.  
- Holdbarhed:  
  - Olie: flere måneder i køleskab, op til 6–12 måneder i fryser  
  - Smør: typisk 2–4 uger i køleskab  

### 7. Dosering / brug
- Start med små mængder (fx 1 tsk) – effekten kan tage **1–2 timer** at mærke.  
- Justér næste portion baseret på ønsket styrke.  

### ✅ Tips for optimal effekt
- **Bland evt. med frisk urt** for at bevare noget klar, energisk high.  
- Lav små batcher først for at teste styrke.  
- Hold øje med smag og lugt – hvis olie eller smør lugter harsk, kassér det.

## 🌿 Vapet urt → Infusion i olie/smør: Effektprofil

### 1. Effekt over tid

| Tidspunkt | THC | THCV | CBN | Ruskarakter |
|-----------|-----|------|-----|-------------|
| Før infusion (vapet urt) | Lav–moderat | Lav | Lille | Mildt energisk → lidt stoner |
| Efter infusion 1–2 timer | Stabil | Stabil | Minimal | Mildere, lettere kropslig stoner |
| Opbevaring 1 uge i olie | Stabil | Stabil | Minimal | Klar rus bevares, ikke tung |
| Opbevaring 2–4 uger | Mildt fald | Mildt fald | Lidt stigende | Mere stoner-lignende, mindre energi |
| 4+ uger | Markant fald | Markant fald | Øget | Kropslig, tung rus, energi næsten væk |

> Bemærkning: Lav temperatur og mørk opbevaring minimerer THC/THCV-tab og CBN-dannelse.

---

### 2. Effekt af blanding med frisk urt

| Forhold (frisk : vapet) | Forventet rus |
|--------------------------|---------------|
| 2:1 | Mest energisk / klar high |
| 1:1 | Blandet / balanceret energi + stoner |
| 1:2 | Mest stoner / tung rus |

---

### 3. Praktiske pointer

- Pres urten efter infusion → maksimer udbyttet af cannabinoider.  
- Opbevar olie/smør køligt og mørkt for maksimal holdbarhed.  
- Brug små portioner først → edibles har langsom onset (1–2 timer).  
- Frisk urt kan bruges sammen med vapet urt for at bevare noget klar, energisk effekt.
