# 🔬 Vaping: Skunk vs. THC(A)-isolatblanding (samme styrke)

| Faktor | 🌿 Skunk (naturlig cannabis) | 💎 Isolatblanding (THCA + THC-fri urt) |
|--------|-----------------------------|---------------------------------------|
| Sammensætning | Indeholder THC, CBD, CBN, CBG, terpener, flavonoider og planteolier | Næsten kun THC (fra isolat) + evt. lidt CBD/terpener fra urten |
| Entourage-effekt | Ja — cannabinoider og terpener arbejder synergistisk (afbalancerer og ændrer rus) | Begrænset — THC virker mere “rent” og direkte |
| High-karakter | Ofte rundere, dybere og mere “kropslig” rus | Mere ren, klar og fokuseret — men også kortere og mere lineær |
| Smag & aroma | Naturligt kompleks (terpener) | Afhænger af CBD-urten — ofte mildere, lidt mere “neutral” |
| CBN-niveau (træthed) | Typisk noget CBN → mere chill/stoney | Næsten intet CBN → mindre tunghed, mere energi |
| Toleranceopbygning | Langsommere pga. cannabinoid-balance | Hurtigere, da THC står “alene” |
| Medicinsk profil | Varierer — nogle strains beroliger, andre stimulerer | Meget kontrollerbar — du styrer helt selv effekten via dosis og temp. |
| Afterglow (efterfølelse) | Længere, men kan gøre dig sløv | Kortere, men mere ren og klar |

## 🧠 Kort sagt:

**Skunk** = kompleks high → “helhedsoplevelse” med synergier.  
**Isolatblanding** = klinisk ren high → hurtigere, klarere, men mindre nuanceret.

## 💬 Hvordan det føles i praksis:

- Skunk føles som en bølge, der både rammer hoved og krop.  
- Isolat føles som et skud ren energi i panden — hurtigt ind, hurtigt ud.

# 🌿💎 Entourage-effekten i isolatblandinger

| Element | Kommer fra | Virkning | Effekt i blandingen |
|---------|------------|----------|-------------------|
| THC / THCA | Isolat | Hovedpsykoaktiv komponent | Primær drivkraft for rus |
| CBD | THC-fri urt (fx hampblomst) | Dæmper angst og overstimulation | Giver mere roligt og klart fokus |
| Terpener | Urten (fx myrcen, limonen, pinene) | Justerer stemning og kropsrespons | Gør high’en mere “naturlig” og behagelig |
| CBG, CBC m.fl. | Sporstoffer i CBD-urten | Små modulerende effekter | Understøtter balance og klarhed |

## 🧪 Samlet:

En blanding af THCA-isolat + CBD- og terpenerig urt fungerer som en syntetisk entourage-effekt — du bygger den selv fra bunden.  

Den bliver ikke helt så kompleks som en naturlig skunk (der har 100+ aktive forbindelser),  
men du får stadig en mærkbar forskel i:  

- ro i kroppen  
- mindre paranoia/overstimulation  
- mere smooth overgang mellem “på” og “ned”

---
| Produkt / Setup | THC-styrke | CBD | Terpener | Entourage-effekt | Bemærkninger |
|-----------------|------------|-----|----------|-----------------|--------------|
| Skunk (naturlig) | Varierende, typisk 15–25% | Lav-moderat | Fuldt spektrum | 🌈 Fuld | Traditionel high, kompleks aroma og virkning |
| THC-isolat alene | 99%+ | 0 | Ingen | ❌ Ingen | Ren psykoaktiv effekt, ingen aroma eller kompleksitet |
| THC-isolat + CBD-urt (5:1) | Ca. 20% | Tilstede via urt | Begrænset | ⚙️ Delvis / begrænset | Minder om naturlig skunk styrkemæssigt, men entourage-effekt er reduceret |

---
## 💬 Kort sagt:

- **Rent isolat** = THC alene → meget direkte high  
- **Isolat + CBD-urt** = light entourage → mere balanceret, socialt og anvendeligt high  
- **Skunk** = fuld entourage → naturligt, men sværere at styre
