# Terpener i cannabis – mere end bare duft

```{admonition} Terpener
:class: note

- De fleste kender til THC og CBD, men få er klar over, hvor stor en rolle **terpener** spiller i cannabisoplevelsen.
- Terpener er aromatiske forbindelser, som giver planten dens karakteristiske **duft og smag** – men deres funktion stopper ikke der. De påvirker også **stemning, energi, kognition og krop** – ofte i tæt samspil med cannabinoiderne.
```

---
## 🌸 Hvad er terpener?

Terpener er en stor gruppe organiske forbindelser, der findes i næsten alle planter, træer og blomster. I naturen beskytter de planter mod skadedyr og tiltrækker bestøvere. Når det kommer til cannabis, **menes terpener at interagere med det endocannabinoide system** og påvirker både nervesystemet og immunsystemet.

Der findes over 100 forskellige terpener i cannabis, men cirka 10–15 af dem er de mest fremtrædende og har størst indflydelse på den samlede effekt. Hver cannabis-strain har sit eget unikke terpenprofil, som kan variere afhængigt af genetisk sammensætning og dyrkningsforhold. Det betyder, at de terpener, der dominerer i en bestemt strain, er med til at forme, hvordan den opleves i krop og sind.

Nogle terpener har en **beroligende effekt**, mens andre virker **stimulerende**, **fokuserende**, **smertelindrende** eller **appetitdæmpende**. De fungerer ofte som "stemningsjusterere", der kan forstærke, balancere eller tone virkningen af THC og andre cannabinoider. For eksempel kan nogle terpener intensivere den beroligende effekt i en indica-strain, mens andre kan fremme en mere energisk og opkvikkende effekt i sativa-strains.

De mest almindelige terpener, du kan finde i cannabis, inkluderer:

- **🌲 Pinene**: Kendt for sin fyrretræsduft og evne til at forbedre mental klarhed og opmærksomhed.
- **🌿 Myrcen**: En jordet, krydret duft, som har en beroligende virkning og kan hjælpe med søvn.
- **🍋 Limonen**: Frisk og citrusagtig, limonen virker opkvikkende og angstdæmpende.
- **💜 Linalool**: Lavendelduftende og beroligende, ofte brugt til at fremme afslapning og reducere stress.
- **🌶️ Caryophyllen**: Krydret og peberagtig, dette terpen binder direkte til CB2-receptorer og har antiinflammatoriske egenskaber.

Kombinationen af disse terpener, sammen med de specifikke cannabinoider i en strain, skaber den unikke effekt, som hver strain tilbyder. Det er derfor, nogle cannabis-strains kan få dig til at føle dig energisk og fokuseret, mens andre giver dig en dyb afslapning eller lindring af smerte.


---

## 🧬 Terpen-DNA: Hvordan duften former oplevelsen

Cannabisplanter kategoriseres ofte som **indica**, **sativa** eller **hybrid** – men disse betegnelser siger ikke nødvendigvis meget om den oplevede virkning. Mange forskere mener, at **terpensammensætningen** er en bedre indikator. Kolonnen **Strain-type** bør derfor tages med et gran salt (eller urt).\
Denne “duftkode” – eller **terpen-DNA** – påvirker både **stemning, energiniveau, mental tilstand og kropslig følelse**.

| Strain-type             | Dominerende terpener                               | Duft / Smag                             | Psykisk effekt                                | Kropslig effekt                          | Samlet stemning                         |
|-------------------------|-----------------------------------------------------|------------------------------------------|------------------------------------------------|-------------------------------------------|------------------------------------------|
| 🌞 **Sativa-domineret**  <br>  | Limonen, Pinene, Terpinolen, Ocimen               | Citrus, fyr, blomster, frisk              | Euforisk, kreativ, social, klarhed            | Let krop, ingen tunghed                   | Lys, energisk, opløftende                |
| 🌙 **Indica-domineret**  <br>  | Myrcen, Linalool, Humulen, Caryophyllen           | Jordet, lavendel, træ, krydderi           | Mental ro, dæmpet tankeaktivitet              | Muskelafslappende, søvndyssende           | Dyb afslapning, “stoner”, varm og blød   |
| ⚖️ **Hybrid-typer**       <br>  | Fx Limonen + Myrcen + Caryophyllen                | Sød, frugtig, krydret                      | Blandet high: energi + introspektion          | Afbalanceret, mild varme                  | Behagelig “balanceret high”              |


---

## 💨 Terpenernes rolle i vaping

Terpener frigives ved forskellige temperaturer, afhængigt af den enkelte forbindelse. Mange af de mest aromatiske terpener begynder at frigives allerede fra omkring **155–175 °C** og bliver gradvist udtømt, efterhånden som temperaturen stiger. Starter man direkte ved en høj temperatur, vil de derfor ofte bidrage mindre til både aroma og den samlede oplevelse.

Valget af temperatur afhænger derfor af, hvad man ønsker at opnå:

- **Omkring 165–180 °C** fremhæver typisk smag, aroma og de mest flygtige terpener.
- **Gradvis temperaturstigning** giver både terpener og cannabinoider mulighed for at bidrage løbende gennem sessionen.
- **Højere temperaturer** giver generelt en kraftigere og mere komplet ekstraktion, men med mindre bidrag fra de mest flygtige terpener.

---

## 🤝 Entourage-effekten: Samspillet mellem terpener og cannabinoider

Flere studier peger på, at terpener ikke blot bidrager med aroma, men også kan påvirke, **hvordan THC og CBD opleves**.\
Denne synergieffekt kaldes **entourage-effekten** – et slags “helhedsudtryk”, hvor flere stoffer arbejder sammen og skaber en samlet oplevelse, der er mere end summen af delene.

Eksempelvis:

- **🌲 Pinene** kan modvirke nogle af THC's hukommelsessvækkende effekter og give mere mental klarhed
- **🌿 Myrcen** menes at øge cellernes optagelse af THC og dermed gøre rusen stærkere og mere kropslig
- **🌶️ Caryophyllen** binder direkte til kroppens CB2-receptorer og kan dæmpe inflammation og uro


---
## 🧪 Terpenoversigt – Frigivelsestemperatur og virkning

Hver terpen har sin egen **aroma, fysiologiske effekt og stemningsmæssige “farve”**.

| **Terpen** | **Frigives fra ca. (°C)** | **Duft / Smag** | **Psykisk virkning** | **Kropslig virkning** |
|------------|--------------------------:|-----------------|----------------------|-----------------------|
| **🌲 Pinene** | ~155 | Fyr, skov, frisk | Klarhed, fokus, opmærksomhed | Anti-inflammatorisk, bronkieudvidende |
| **🌿 Myrcen** | ~165 | Jordet, muskat, let krydret | Beroligende, søvnfremmende | Muskelafslappende, forstærker THC's “stoner”-effekt |
| **🍋 Limonen** | ~175 | Citrus, frisk, sød | Eufori, humørløftende | Anti-angst, stressdæmpende |
| **🌸 Terpinolen** | ~185 | Frisk, blomster, let krydret | Opløftende, energisk | Antioxidant |
| **🌷 Ocimen** | ~190 | Sød, blomster | Stimulerende, kreativ | Antiviral, aromaforstærkende |
| **💜 Linalool** | ~200 | Lavendel, blomster | Afslappende, angstdæmpende | Smertelindrende, søvnfremmende |
| **🌶️ Caryophyllen** | ~200 | Peber, krydderi | Afbalancerende, modvirker stress | Binder til CB2-receptorer, anti-inflammatorisk |
| **🍺 Humulen** | ~210 | Humle, jordet, lidt bitter | Mildt beroligende | Appetitdæmpende, antibakteriel |
| **🧴 Bisabolol** | ~220 | Mild, blomster | Beroligende, rolig | Hudvenlig, anti-inflammatorisk |

---

💡 **Sådan bruges tabellen**  
Brug denne oversigt som reference til **at vælge terpener** ud fra ønsket stemning og virkning. Temperaturerne angiver de omtrentlige temperaturer, hvor de enkelte terpener typisk **begynder at frigives**. Ved højere temperaturer kan de fortsat bidrage til oplevelsen, men deres relative betydning ændrer sig gradvist, efterhånden som de frigives og udtømmes.

```{dropdown} 🌸 Terpener – uddybende beskrivelser
:icon: flask
:color: info

**🌲 Pinene**
- Frigives fra ca. **~155 °C**
- Duft: Fyrretræ, harpiks, skov
- **Effekt:**
  - Forbedrer mental klarhed og opmærksomhed
  - Kan **modvirke hukommelsessvækkelse** forårsaget af THC
  - Har **anti-inflammatoriske egenskaber** og virker luftvejsudvidende

---

**🌿 Myrcen**
- Frigives fra ca. **~165 °C**
- Duft: Jordet, krydret, moskus
- **Effekt:**
  - **Forstærker THC's rus** og gør den mere kropslig
  - Har beroligende, søvnfremmende og muskelafslappende virkning
  - Myrcen menes at **øge cellemembranernes optagelse af cannabinoider**

---

**🍋 Limonen**
- Frigives fra ca. **~175 °C**
- Duft: Citronskal, lime, appelsin
- **Effekt:**
  - Løfter humør, giver energi og motivation
  - Kan virke **angstdæmpende og stressreducerende**
  - Øger optagelsen af andre terpener gennem huden

---

**🌸 Terpinolen**
- Frigives fra ca. **~185 °C**
- Duft: Frisk, let krydret, blomsteragtig
- **Effekt:**
  - Mildt **stimulerende** og antioxidant
  - Mindre udbredt i rene strains, men vigtig i kombination med limonen og pinene
  - Understøtter sativa-lignende terpeneprofiler

---

**🌷 Ocimen**
- Frigives fra ca. **~190 °C**
- Duft: Sød, floral, urteagtig
- **Effekt:**
  - Har **antiviral og svampedræbende virkning**
  - Giver en let og aromatisk stemning
  - Bruges ofte i parfume og æteriske olier

---

**💜 Linalool**
- Frigives fra ca. **~200 °C**
- Duft: Lavendel, blomster
- **Effekt:**
  - **Søvndyssende og angstdæmpende**, bruges også i aromaterapi
  - Virker mildt smertestillende
  - Kan støtte kroppens reaktion på stress

---

**🌶️ Caryophyllen**
- Frigives fra ca. **~200 °C**
- Duft: Peber, træ, krydderi
- **Effekt:**
  - **Binder direkte til CB2-receptorer** (unik blandt terpener)
  - **Dæmper inflammation, uro og fysisk ubehag**
  - Bruges terapeutisk mod angst og irritabel tarm

---

**🍺 Humulen**
- Frigives fra ca. **~210 °C**
- Duft: Humle, jord, bitterhed
- **Effekt:**
  - Mildt beroligende og **appetitdæmpende**
  - Har antibakterielle egenskaber
  - Samme terpen, som giver humle dens karakteristiske aroma

---

**🧴 Bisabolol**
- Frigives fra ca. **~220 °C**
- Duft: Mild, blomsteragtig (kamilleagtig)
- **Effekt:**
  - Virker **beroligende og anti-inflammatorisk**
  - Meget **hudvenlig**, bruges i salver og kosmetik
  - Kombineres ofte med linalool i beroligende terpeneprofiler
```

## 💡 Sådan kan du bruge det

Når du forstår terpenernes funktion og samspil med cannabinoider, kan du **justere din oplevelse** mere præcist.  
Her er nogle enkle retningslinjer til at vælge ud fra ønsket effekt:

- 🌞 **Vil du have kreativ energi og social klarhed?**  
  → Vælg sativa-dominerede strains med **limonen**, **pinene** og **terpinolen**.

- 🌙 **Vil du slappe af eller falde i søvn?**  
  → Gå efter indica-profiler med **myrcen**, **linalool** og **caryophyllen**.

- ⚖️ **Vil du balancere krop og sind?**  
  → Hybridtyper med **limonen** og **caryophyllen** giver ofte en rolig, men opløftende high.

Ved at justere både **temperatur** og **terpenprofil** i din session, kan du skabe en oplevelse, der er langt mere **målrettet og behagelig** – uanset om du søger klarhed, energi, afslapning eller søvn.
